```python
# Install required libraries if needed
# pip install networkx matplotlib

import networkx as nx
import matplotlib.pyplot as plt

# Step 1: Create an empty undirected graph
G = nx.Graph()

# Step 2: Add nodes (can be any hashable object: numbers, strings, etc.)
G.add_nodes_from([1, 2, 3, 4, 5])

# Step 3: Add edges (unordered pairs since it's undirected)
G.add_edges_from([
    (1, 2),
    (1, 3),
    (2, 4),
    (3, 4),
    (4, 5)
])

# Step 4: Draw the graph
plt.figure(figsize=(6, 4))
nx.draw(
    G,
    with_labels=True,
    node_color='skyblue',
    node_size=1000,
    edge_color='gray',
    font_weight='bold'
)
plt.title("Dummy Undirected Graph")
plt.show()
```
